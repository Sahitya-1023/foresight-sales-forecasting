```python
# ============================================================
# CATEGORY ANALYSIS
# ============================================================

print("=" * 60)
print("CATEGORY ANALYSIS")
print("=" * 60)

# Get all daily sales columns
sales_cols = [col for col in sales.columns if col.startswith("d_")]

# Calculate total sales for each item
sales["total_sales"] = sales[sales_cols].sum(axis=1)

# Total sales by category
category_sales = (
    sales.groupby("cat_id")["total_sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTotal Sales by Category:")
print(category_sales)

# Percentage contribution of each category
category_percentage = (
    category_sales / category_sales.sum() * 100
).round(2)

print("\nCategory Sales Percentage:")
print(category_percentage)
```

```python
# ============================================================
# STEP 9: MONTHLY SEASONALITY
# ============================================================

print("=" * 60)
print("MONTHLY SEASONALITY")
print("=" * 60)

# Load calendar dataset
calendar = pd.read_csv("../data/Raw/calendar.csv")

print("\nCalendar shape:", calendar.shape)
print("Calendar columns:")
print(calendar.columns.tolist())
```

```python
# ============================================================
# STEP 9: MONTHLY SEASONALITY
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

print("=" * 60)
print("MONTHLY SEASONALITY")
print("=" * 60)

# Make sure date is in datetime format
calendar["date"] = pd.to_datetime(calendar["date"])

# Get only the daily sales columns
sales_cols = [col for col in sales.columns if col.startswith("d_")]

# Total sales for each day across all products
daily_sales = sales[sales_cols].sum(axis=0)

# Create daily sales dataframe
daily_sales_df = pd.DataFrame({
    "d": daily_sales.index,
    "total_sales": daily_sales.values
})

# ------------------------------------------------------------
# M5 calendar does not contain a 'd' column.
# Create d_1, d_2, ..., d_1913 based on calendar row order.
# ------------------------------------------------------------

calendar_m5 = calendar.iloc[:len(sales_cols)].copy()

calendar_m5["d"] = sales_cols

# Merge sales with calendar
monthly_data = daily_sales_df.merge(
    calendar_m5[["d", "date", "month", "year"]],
    on="d",
    how="left"
)

# Calculate total sales by month
monthly_sales = (
    monthly_data
    .groupby(["year", "month"])["total_sales"]
    .sum()
    .reset_index()
)

print("\nMonthly Sales:")
print(monthly_sales)

print("\nNumber of months:", len(monthly_sales))
```

```python
# ============================================================
# MONTHLY SEASONALITY VISUALIZATION
# ============================================================

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_data["date"].drop_duplicates(),
    monthly_data.groupby("date")["total_sales"].sum()
)

plt.title("Monthly Sales Seasonality - M5 Dataset")
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# STEP 10: AVERAGE SALES BY MONTH
# ============================================================

print("=" * 60)
print("AVERAGE SALES BY MONTH")
print("=" * 60)

# Calculate average sales for each calendar month
average_monthly_sales = (
    monthly_data
    .groupby("month")["total_sales"]
    .mean()
)

print("\nAverage Sales by Month:")
print(average_monthly_sales)

# Create month names
month_names = {
    1: "January",
    2: "February",
    3: "March",
    4: "April",
    5: "May",
    6: "June",
    7: "July",
    8: "August",
    9: "September",
    10: "October",
    11: "November",
    12: "December"
}

average_monthly_sales.index = average_monthly_sales.index.map(month_names)

print("\nAverage Sales by Month:")
print(average_monthly_sales.round(2))
```

```python
# ============================================================
# STEP 11: RISK ANALYSIS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("=" * 60)
print("RISK ANALYSIS")
print("=" * 60)

# ------------------------------------------------------------
# 1. Prepare item-level sales data
# ------------------------------------------------------------

sales_cols = [col for col in sales.columns if col.startswith("d_")]

risk_data = pd.DataFrame()

risk_data["item_id"] = sales["item_id"]
risk_data["total_sales"] = sales[sales_cols].sum(axis=1)
risk_data["average_sales"] = sales[sales_cols].mean(axis=1)
risk_data["std_sales"] = sales[sales_cols].std(axis=1)

# ------------------------------------------------------------
# 2. Calculate coefficient of variation
# ------------------------------------------------------------

risk_data["cv"] = np.where(
    risk_data["average_sales"] > 0,
    risk_data["std_sales"] / risk_data["average_sales"],
    np.nan
)

# ------------------------------------------------------------
# 3. Calculate zero-sales percentage
# ------------------------------------------------------------

risk_data["zero_sales_percentage"] = (
    (sales[sales_cols] == 0).sum(axis=1)
    / len(sales_cols)
) * 100

# ------------------------------------------------------------
# 4. Display risk statistics
# ------------------------------------------------------------

print("\nRisk Statistics:")
print("-" * 60)

print("Average demand:",
      round(risk_data["average_sales"].mean(), 2))

print("Average demand variability:",
      round(risk_data["std_sales"].mean(), 2))

print("Average coefficient of variation:",
      round(risk_data["cv"].mean(), 2))

print("Average zero-sales percentage:",
      round(risk_data["zero_sales_percentage"].mean(), 2), "%")

# ------------------------------------------------------------
# 5. Identify high-volatility items
# ------------------------------------------------------------

high_volatility = risk_data[
    risk_data["cv"] > risk_data["cv"].quantile(0.75)
]

print("\nHigh-volatility items:",
      len(high_volatility))

# ------------------------------------------------------------
# 6. Identify high zero-sales items
# ------------------------------------------------------------

high_zero_sales = risk_data[
    risk_data["zero_sales_percentage"] >
    risk_data["zero_sales_percentage"].quantile(0.75)
]

print("High zero-sales items:",
      len(high_zero_sales))

# ------------------------------------------------------------
# 7. Risk classification
# ------------------------------------------------------------

def classify_risk(row):

    if row["cv"] >= 1.0 or row["zero_sales_percentage"] >= 50:
        return "High Risk"

    elif row["cv"] >= 0.5 or row["zero_sales_percentage"] >= 25:
        return "Medium Risk"

    else:
        return "Low Risk"


risk_data["risk_level"] = risk_data.apply(
    classify_risk,
    axis=1
)

# ------------------------------------------------------------
# 8. Risk distribution
# ------------------------------------------------------------

risk_distribution = risk_data["risk_level"].value_counts()

print("\nRisk Distribution:")
print("-" * 60)
print(risk_distribution)

# ------------------------------------------------------------
# 9. Risk percentage
# ------------------------------------------------------------

risk_percentage = (
    risk_distribution /
    len(risk_data)
) * 100

print("\nRisk Percentage:")
print("-" * 60)
print(risk_percentage.round(2))

# ------------------------------------------------------------
# 10. Risk visualization
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

risk_distribution.reindex(
    ["Low Risk", "Medium Risk", "High Risk"]
).plot(kind="bar")

plt.title("Demand Risk Distribution - M5 Dataset")
plt.xlabel("Risk Level")
plt.ylabel("Number of Items")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
